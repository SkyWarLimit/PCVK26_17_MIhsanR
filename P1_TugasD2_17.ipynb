**Langkah 1**

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow #untuk display image
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

**Langkah 2**

```python
#buat list untuk nyimpen url dari beberapa gambar
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg"]
#loop tiap url image, beberapa dapat disimpan di list
for url in urls:
  image = io.imread(url) #membaca gambar
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize gambar
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert gambar ke RGB
  final_frame = cv.hconcat((image, image_2)) #menggabungkan gambar
  cv2_imshow(final_frame) #menampilkan gambar
  print('\n')
```

Pada langkah 2 Foto ketiga (foto kucing) tidak bisa dibuka karna dari link aslinya error, sehingga penyelesaiannya hanya menggunakan foto **pertama** dan **kedua**

**Langkah 3**

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi , " x ", lebar)
cv2_imshow(image_2)
```

**Langkah 4**

```python
image_3 = image_2.copy() #mendefinisikan image 3 adalah copy dari image 2

#buat garis horizontal ditengah gambar
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

karna pada gambar ketiga tidak bisa dimasukkan, maka langkah 4 diganti dengan menggunakan gambar kedua

**Tugas Praktikum D2**
1. membuat garis vertikal dan garis menyilang diagonal
2. membuat garis horisontal berwarna putih dibagian tengah gambar dengan panjang tertentu
3. buat kotak menggunakan kumpulan pixel warna putih di sembarang tempat dalam gambar

```python
image_4 = image.copy()
for x in range (tinggi):
  image_4[x, int((lebar)/2)] = [255,255,255]
final_frame = cv.hconcat((image, image_4))
print("garis vertikal di tengah gambar")
cv2_imshow(final_frame)

```

```python
image_5_silang = image_2.copy()

for y in range (tinggi):
  x1 = int(y * (lebar/tinggi))
  x2 = int((tinggi - 1 - y) * (lebar/tinggi))

  x1 = min(x1, lebar - 1)
  x2 = min(x2, lebar - 1)
  image_5_silang[y, x1] = [255,255,255]
  image_5_silang[y, x2] = [255,255,255]
final_frame = cv.hconcat((image_2, image_5_silang))
print("1. garis diagonal berwarna putih")
cv2_imshow(final_frame)
```

```python
image_6 = image_2.copy()
panjang_garis = 100
y_tengah = tinggi // 2
x_tengah = lebar // 2
x_awal = max(0, x_tengah - (panjang_garis // 2))
x_akhir = min(lebar, x_tengah + (panjang_garis // 2))


for x in range(x_awal, x_akhir):
    image_6[y_tengah, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_6))
print("2. garis horisontal berwarna putih di tengah gambar")
cv2_imshow(final_frame)
```

```python
image_7 = image_2.copy()
ukuran_kotak = 100
x_awal = 150
x_akhir = x_awal + ukuran_kotak

y_awal = 50
y_akhir = y_awal + ukuran_kotak

for y in range(y_awal, y_akhir):
    for x in range(x_awal, x_akhir):
        image_7[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_7))
print("3. kotak menggunakan kumpulan pixel warna putih di sembarang tempat")
cv2_imshow(final_frame)
```
