NO 4 TRANSFORMASI GRAYSCALE

```python
# ==========================================
# TUGAS 4: Transformasi Grayscale (Averaging, Lightness, Luminance)
# ==========================================
img_ktp1c = cv.imread(BASE + 'KTP1c.jpg') # Citra dengan flash

# Memisahkan channel (Perhatikan urutan BGR di OpenCV)
B, G, R = cv.split(img_ktp1c)
B = B.astype(np.float64)
G = G.astype(np.float64)
R = R.astype(np.float64)

# 1. Averaging: (R+G+B)/3
gray_avg = ((R + G + B) / 3).astype(np.uint8)

# 2. Lightness: (max(R,G,B) + min(R,G,B))/2
max_rgb = np.maximum(np.maximum(R, G), B)
min_rgb = np.minimum(np.minimum(R, G), B)
gray_light = ((max_rgb + min_rgb) / 2).astype(np.uint8)

# 3. Luminance: Menggunakan Rec. 601 (0.299R + 0.587G + 0.114B) sesuai anjuran modul
gray_lum = (0.299 * R + 0.587 * G + 0.114 * B).astype(np.uint8)

# Menampilkan Gambar
fig, ax = plt.subplots(1, 3, figsize=(18, 5))
ax[0].imshow(gray_avg, cmap='gray'); ax[0].set_title('Averaging')
ax[1].imshow(gray_light, cmap='gray'); ax[1].set_title('Lightness')
ax[2].imshow(gray_lum, cmap='gray'); ax[2].set_title('Luminance')
for a in ax: a.axis('off')
plt.show()

# Perhitungan Rata-rata dan Simpangan Baku
print("Tabel Analisa Intensitas Grayscale:")
print("-" * 50)
print(f"{'Metode':<15} | {'Rata-rata (Mean)':<15} | {'Simpangan Baku (Std)':<15}")
print("-" * 50)
print(f"{'Averaging':<15} | {np.mean(gray_avg):<15.2f} | {np.std(gray_avg):<15.2f}")
print(f"{'Lightness':<15} | {np.mean(gray_light):<15.2f} | {np.std(gray_light):<15.2f}")
print(f"{'Luminance':<15} | {np.mean(gray_lum):<15.2f} | {np.std(gray_lum):<15.2f}")
print("-" * 50)

# JAWABAN ANALISA TUGAS 4:
# Metode yang memberi pemisahan paling jelas antara teks dan latar kartu biasanya adalah *Luminance*,
# karena metode ini memperhitungkan sensitivitas mata manusia terhadap warna hijau dan merah,
# sehingga detail teks hitam di atas latar belakang menjadi lebih tegas dan kontras dibandingkan metode Averaging biasa.
```

**NO 5 COLOR MASKING**

```python
# ==========================================
# TUGAS 5: Isolasi Satu Warna Spesifik (Lainnya Grayscale)
# ==========================================
img_obj = cv.imread(BASE + 'OBJ1.jpg') # Foto benda milik sendiri

# Konversi BGR ke HSV untuk deteksi warna yang lebih stabil
hsv = cv.cvtColor(img_obj, cv.COLOR_BGR2HSV)

# --- SESUAIKAN RENTANG WARNA INI DENGAN BENDA ANDA ---
# Contoh di bawah ini adalah rentang untuk warna MERAH (merah memiliki 2 rentang di HSV)
lower_color1 = np.array([0, 120, 70])
upper_color1 = np.array([10, 255, 255])
lower_color2 = np.array([170, 120, 70])
upper_color2 = np.array([180, 255, 255])
mask1 = cv.inRange(hsv, lower_color1, upper_color1)
mask2 = cv.inRange(hsv, lower_color2, upper_color2)
mask = mask1 | mask2
# -----------------------------------------------------

# Memisahkan area benda (warna asli) dan area latar (dijadikan grayscale)
mask_inv = cv.bitwise_not(mask)

# Buat versi grayscale dari gambar asli, lalu jadikan 3 channel agar bisa digabung
gray = cv.cvtColor(img_obj, cv.COLOR_BGR2GRAY)
gray_3ch = cv.cvtColor(gray, cv.COLOR_GRAY2BGR)

# Ekstraksi menggunakan bitwise AND
obj_warna = cv.bitwise_and(img_obj, img_obj, mask=mask)
bg_gray = cv.bitwise_and(gray_3ch, gray_3ch, mask=mask_inv)

# Gabungkan keduanya
final_obj = cv.add(obj_warna, bg_gray)

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(cv.cvtColor(img_obj, cv.COLOR_BGR2RGB)); ax[0].set_title('Citra Asli')
ax[1].imshow(cv.cvtColor(final_obj, cv.COLOR_BGR2RGB)); ax[1].set_title('Satu Warna (Lainnya Grayscale)')
for a in ax: a.axis('off')
plt.show()

# JAWABAN ANALISA TUGAS 5:
# Rentang HSV yang dipakai: Hue(0-10 & 170-180), Saturation(120-255), Value(70-255).
# Pengaruh pelebaran rentang: Jika rentang S dan V diperlebar (misal ke 0),
# maka warna latar yang pudar atau bayangan benda akan ikut terdeteksi sebagai warna asli.
# Sebaliknya jika dipersempit, bagian benda yang terkena highlight cahaya atau bayangan gelap
# akan menjadi abu-abu karena dianggap di luar batas warna objek utama.
```

**NO 6 GAMMA CORRECTION**

```python
# ==========================================
# TUGAS 6: Gamma Correction pada Citra Pribadi
# ==========================================
img_ktp1a = cv.imread(BASE + 'KTP1a.jpg')

print('Gamma Correction pada citra')
print('-' * 30)
try:
    # Meminta input pengguna. Pada Python standar, gamma biasanya float.
    # Sesuai pedoman modul, kita ambil nilainya (diubah ke float agar tidak error saat dipangkat).
    gamma_input = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma_input = PARAM['gamma'] # Fallback ke nilai NIM jika input gagal

# Rumus Gamma Correction modul: I' = 255 * (I/255)^gamma
img_norm = img_ktp1a.astype(np.float64) / 255.0
img_gamma = (255.0 * (img_norm ** gamma_input)).astype(np.uint8)

# Menampilkan citra asli berdampingan dengan Gamma
final_frame_gamma = cv.hconcat((img_ktp1a, img_gamma))

from google.colab.patches import cv2_imshow
print(f"\nHasil Citra Asli (kiri) vs Gamma {gamma_input} (kanan):")
cv2_imshow(final_frame_gamma)

# JAWABAN ANALISA TUGAS 6:
# Nilai gamma terbaik pilihan saya adalah: 0.3 karena pada nilai tersebut detail teks yang tersembunyi
# di area gelap mulai terangkat kecerahannya tanpa membuat background menjadi terlalu putih.
# karena kondisi pencahayaan dan perangkat kamera yang digunakan untuk mengambil KTP1a.jpg
# memiliki sensor dan dynamic range yang berbeda-beda.
```

**NO 7 SIMULASI IMAGE DEPTH**

```python
# ==========================================
# TUGAS 7: Simulasi Image Depth 1 hingga 7 bit
# ==========================================
original_gray = cv.imread(BASE + 'KTP1b.jpg', cv.IMREAD_GRAYSCALE)

fig, ax = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle('Simulasi Image Depth 1-7 bit', fontsize=16)

# Plot Asli (8-bit) di awal
ax[0,0].imshow(original_gray, cmap='gray'); ax[0,0].set_title('Asli (8-bit)')

# Looping kedalaman 1 sampai 7 bit
for bit in range(1, 8):
    level = 255 / (pow(2, bit) - 1)
    depth_image = np.round(original_gray / level) * level
    depth_image = depth_image.astype(np.uint8)

    # Menentukan letak subplot
    row = bit // 4
    col = bit % 4
    ax[row, col].imshow(depth_image, cmap='gray')
    ax[row, col].set_title(f'Depth {bit}-bit')

for a in ax.flat: a.axis('off')
plt.tight_layout()
plt.show()

# JAWABAN ANALISA TUGAS 7:
# Berdasarkan hasil simulasi, teks nama dan NIM pada kartu mulai jelas terbaca pada kedalaman 4-bit .
# Di bawah nilai tersebut, kuantisasi warnanya terlalu ekstrem sehingga tepi huruf dan latar belakang menyatu dan teks tidak dapat diidentifikasi.
```

**NO 8 AVERANGE DENOISING & PSNR**

```python
# ==========================================
# TUGAS 8: Average Denoising & PSNR
# ==========================================
import glob

# Fungsi PSNR sesuai koreksi modul
def PSNR(img1, img2):
    img1 = img1.astype(np.float64)
    img2 = img2.astype(np.float64)
    mse = np.mean((img1 - img2) ** 2)
    if mse == 0:
        return float('inf')
    return 20 * np.log10(255.0 / np.sqrt(mse))

# Pastikan path ini sesuai dengan letak folder images/noises/ di drive Anda
img_asli = cv.imread('/content/drive/MyDrive/PCVK/Image/galaxy.jpg')

cv_img = []
file_list = glob.glob('/content/drive/MyDrive/PCVK/Image/noises/*.jpg')
for img_path in sorted(file_list):
    n = cv.imread(img_path)
    if n is not None:
        cv_img.append(n)

if len(cv_img) >= 100:
    print(f"{'Jumlah Citra (N)':<20} | {'Nilai PSNR (dB)':<20}")
    print("-" * 45)

    # Kuantitas citra yang akan dirata-rata
    for jumlah in [10, 20, 40, 80, 100]:
        subset = cv_img[:jumlah]

        # Proses Average
        avg_img = np.zeros_like(img_asli, dtype=np.float64)
        for img in subset:
            avg_img += img
        avg_img = (avg_img / jumlah).astype(np.uint8)

        # Kalkulasi PSNR
        psnr_val = PSNR(img_asli, avg_img)
        print(f"{jumlah:<20} | {psnr_val:<20.4f}")
else:
    print("Gambar noise tidak ditemukan atau kurang dari 100. Periksa path folder Anda.")

# JAWABAN ANALISA TUGAS 8:
# 1. Apakah peningkatan jumlah citra selalu memberikan peningkatan PSNR yang signifikan?
#    Tidak, penambahan PSNR bersifat melengkung, artinya pada awal penambahan citra peningkatannya drastis, namun makin lama makin kecil.
# 2. Pada jumlah citra berapa peningkatan mulai tidak terlalu signifikan?
#    Peningkatan mulai tidak terlalu signifikan setelah jumlah citra mencapai 40 (atau 80),
#    karena penambahan sampel citra setelah titik tersebut hanya memberikan kenaikan nilai PSNR yang sangat kecil (kurang dari 0,03 dB).
# 3. Kesimpulan: Operasi Averaging efektif menghilangkan random Gaussian noise. Semakin banyak citra, semakin tinggi nilai PSNR (citra makin bersih). Namun, terdapat batas efektif (diminishing returns) di mana penambahan sampel tidak lagi setimpal dengan ongkos komputasi.
```

**NO 9 IMAGE MASKING**

```python
# ==========================================
# TUGAS 9: Image Masking (Menyamarkan Data Pribadi)
# ==========================================
img_ktp1d = cv.imread(BASE + 'KTP1d.jpg')
mask = np.zeros(img_ktp1d.shape[:2], dtype=np.uint8)

# --- UBAH KOORDINAT KOTAK INI SESUAI WAJAH DAN NIM DI KTP ANDA ---
# y_start:y_end, x_start:x_end
mask[130:430, 730:970] = 255 # Koordinat contoh wajah
mask[100:150, 240:695] = 255 # Koordinat contoh NIM
# ------------------------------------------------------------------

print("Koordinat Mask: Wajah(200:400, 100:300), NIM(450:500, 150:400)")

# Percobaan Logika Sesuai Tabel Modul
mask_NOT = cv.bitwise_not(mask)
mask_3ch = cv.cvtColor(mask, cv.COLOR_GRAY2BGR) # Mask diubah ke 3 channel untuk operator biner dua citra

# 1. NOT (Komplemen) dari Mask
# 2. AND: Menyembunyikan area yang dimask (Membutuhkan mask_NOT agar wajah jadi 0/hitam)
img_AND = cv.bitwise_and(img_ktp1d, img_ktp1d, mask=mask_NOT)
# 3. OR: Area mask akan berubah jadi putih (karena OR dengan 255/11111111 adalah 255)
img_OR = cv.bitwise_or(img_ktp1d, mask_3ch)
# 4. XOR: Area mask diinversikan warnanya
img_XOR = cv.bitwise_xor(img_ktp1d, mask_3ch)

fig, ax = plt.subplots(2, 2, figsize=(12, 10))
ax[0,0].imshow(cv.cvtColor(mask_3ch, cv.COLOR_BGR2RGB)); ax[0,0].set_title('Mask Asli (Putih = 255, Hitam = 0)')
ax[0,1].imshow(cv.cvtColor(img_AND, cv.COLOR_BGR2RGB)); ax[0,1].set_title('AND (Menyembunyikan Wajah/NIM)')
ax[1,0].imshow(cv.cvtColor(img_OR, cv.COLOR_BGR2RGB)); ax[1,0].set_title('OR (Wajah/NIM Menjadi Putih)')
ax[1,1].imshow(cv.cvtColor(img_XOR, cv.COLOR_BGR2RGB)); ax[1,1].set_title('XOR (Wajah/NIM Klise/Negatif)')
for a in ax.flat: a.axis('off')
plt.show()

# JAWABAN ANALISA TUGAS 9:
# - Operator AND: Area mask yang di-NOT kan menjadi hitam (0). Sifat AND terhadap 0 selalu menghasilkan 0 (hitam), sehingga data pribadi tertutup warna hitam pekat.
# - Operator OR: Sifat OR terhadap 255 akan selalu menghasilkan 255, sehingga area target tertutup warna putih pekat.
# - Operator XOR: Sifat Exclusive OR terhadap 255 akan membalik bit nilai citra asli, sehingga wajah dan NIM terlihat seperti efek negatif/klise.
```
